```python
inputs = [1.0,2.0,3.0,4.0]

weights = [
        [0.2,0.8,-0.5,1],
        [0.5,-0.91,0.26,-0.5],
        [-0.26,-0.27,0.17,0.87],
]
bias = [2,3,0.5]

output = [
    # Neuron 1
      inputs[0]*weights[0][0] + inputs[1]*weights[0][1] + inputs[2]*weights[0][2] + inputs[3]*weights[0][3] + bias[0],
    # Neuron 2
      inputs[0]*weights[1][0] + inputs[1]*weights[1][1] + inputs[2]*weights[1][2] + inputs[3]*weights[1][3] + bias[1],
    # Neuron 3
      inputs[0]*weights[2][0] + inputs[1]*weights[2][1] + inputs[2]*weights[2][2] + inputs[3]*weights[2][3] + bias[2]
]
print(output[0])
print(output[1])
print(output[2])
```

# Improved Code

```python
output = []

for neuron_weights , neuron_bias in zip(weights,bias):
  neuron_output = 0
  for input_value , weight in zip(inputs ,neuron_weights ):
      neuron_output += input_value * weight
  neuron_output += neuron_bias
  output.append(neuron_output)
print(output)
```

```python
inputs = [1,2,3,4]

weights = [
          [0.2,0.8,-0.5,1],
          [0.5, -0.91, 0.26, -0.5],
          [-0.26, -0.27, 0.17, 0.87],
]
```

- Neuron 1 → [0.2, 0.8, -0.5, 1]
- Neuron 2 → [0.5, -0.91, 0.26, -0.5]
- Neuron 3 → [-0.26, -0.27, 0.17, 0.87]

- Each neuron needs 4 weights becz we have 4 inputs
- Each neuron has one bias : bias = [2, 3, 0.5]

```python
neuron_output = []

for neuron_weights, neuron_bias in zip(weights, bias):

    res = 0

    for n_inputs, n_weights in zip(inputs, neuron_weights):
        res += n_inputs * n_weights

    res += neuron_bias
    neuron_output.append(res)

print(f"The neuron output is: {neuron_output[0]}")
print(f"The neuron output is: {neuron_output[1]}")
print(f"The neuron output is: {neuron_output[2]}")
```

# Coding a Neuron using NumPy

```python
import numpy as np

inputs = [1.0,2.0,3.0]
weights = [0.2,0.8,-0.5]
bias = 1

result = np.dot(inputs,weights) + bias
print(result)
```

## *Coding a batch of Data*

```python
inputs = [
      [1.0 ,2.0 , 3.0, 2.5],
      [2,5,-1,2.0],
      [-1.5,2.7,3.3,-0.8]
]
weights = [
    [0.2,0.8,-0.5,1],
    [0.5,-0.91,0.26,-0.5],
    [-0.26 , -0.27 , 0.17 , 0.87]

]
biases = [2.0,3.0,0.5]

outputs = np.dot(inputs,np.array(weights).T)+biases
print(outputs)
```
