## Coding Our First Neuron

![NN](https://drive.google.com/uc?export=view&id=1iSeQV1QfFfA2ESRP33NyR3t-LIZ_V4HF)

```python
inputs = [1.0,2.0,3.0]
weights = [0.2,0.8,-0.5]
bias = 1
outputs = [
            inputs[0]* weights[0] + inputs[1]* weights[1] + inputs[2]* weights[2]  + bias
           ]
print(outputs)
```

```python
inputs = [1.0,2.0,3.0,4.0]
weights = [0.2,0.8,-0.5,1.0]
bias = 2
outputs = [
           inputs[0]*weights[0] + inputs[1]*weights[1] + inputs[2]*weights[2] + inputs[3]*weights[3] + bias
]
print(outputs)
```

```python
inputs = [1.0,2.0,3.0,4.0]

weights = [
        [0.2,0.8,-0.5,1],
        [0.5,-0.91,0.26,-0.5],
        [-0.26,-0.27,0.17,0.87],
]
bias = [2,3,0.5]

output = [
    # Neuron 1
      inputs[0]*weights[0][0] + inputs[1]*weights[0][1] + inputs[2]*weights[0][2] + inputs[3]*weights[0][3] + bias[0],
    # Neuron 2
      inputs[0]*weights[1][0] + inputs[1]*weights[1][1] + inputs[2]*weights[1][2] + inputs[3]*weights[1][3] + bias[1],
    # Neuron 3
      inputs[0]*weights[2][0] + inputs[1]*weights[2][1] + inputs[2]*weights[2][2] + inputs[3]*weights[2][3] + bias[2]
]
print(output[0])
print(output[1])
print(output[2])
```

# Improved Code

```python
output = []

for neuron_weights , neuron_bias in zip(weights,bias):
  neuron_output = 0
  for input_value , weight in zip(inputs ,neuron_weights ):
      neuron_output += input_value * weight
  neuron_output += neuron_bias
  output.append(neuron_output)
print(output)
```
